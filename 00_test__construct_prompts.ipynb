```python
from itertools import product
import yaml
from src.prompts import *
```

```python
S = ['North', 'South', 'West', 'East']
```

# Natural prompts

## 1 and 2

```python
fp = './data/config/prompts_natural_1_2.yaml'
with open(fp, mode='rb') as f:
    data = yaml.safe_load(f)
```

```python
prefix = data['prefix']

templates = data['templates']

features = data['features']
features_grid = [dict(zip(features.keys(), fs)) for fs in product(*features.values())]
```

```python
for template, feature in product(templates, features_grid):
    prompt = create_prompt_natural(template, feature, S, S[0])
    prompt = prefix + ' ' + prompt
    print(prompt + '\n')
```
